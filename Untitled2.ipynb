```python
##Python Function

def hello_world():
    print('Hello, World!')


#return
def add(x, y):
    print("x is %s, y is %s" %(x, y))
    return x + y

add(5, 6)    # => 11

#positional argument
def varargs(*args):
    return args

varargs(1, 2, 3)  # => (1, 2, 3)

#keywords arguments
def keyword_args(**kwargs):
    return kwargs

# => {"big": "foot", "loch": "ness"}
keyword_args(big="foot", loch="ness")

#returning multiple

def swap(x, y):
    return y, x

x = 1
y = 2
x, y = swap(x, y)  # => x = 2, y = 1

#Default value
def add(x, y=10):
    return x + y

add(5)      # => 15
add(5, 20)  # => 25

#Anonymous function

# => True
(lambda x: x > 2)(3)

# => 5
(lambda x, y: x ** 2 + y ** 2)(2, 1)

```
